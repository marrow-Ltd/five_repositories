```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("college.db")
cursor = conn.cursor()
```

```python
def run(query):
    display(pd.read_sql_query(query, conn))
```

```python
cursor.execute("DROP TABLE IF EXISTS Marks")

cursor.execute("""
CREATE TABLE Marks(
    MarkID INTEGER PRIMARY KEY,
    StudentID INTEGER,
    Subject TEXT,
    Marks INTEGER,
    Semester INTEGER,
    FOREIGN KEY(StudentID) REFERENCES Students(StudentID)
)
""")

conn.commit()
```
